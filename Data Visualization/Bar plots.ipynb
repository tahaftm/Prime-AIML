```python
import matplotlib.pyplot as plt
```

### Creating side by side bar plots

```python
import numpy as np
names = ['Sara','Ali','Danish','Arhum']
math_marks = [56,78,89,87]
english_marks = [78,98,87,66]

x = np.arange(len(names))
width = 0.4

plt.bar(x-width/2,math_marks,width)
plt.bar(x+width/2,english_marks,width)

plt.xticks(x, names)

plt.title("Marks Dsitribution")
plt.xlabel("Names")
plt.ylabel("marks")
plt.show()
```
