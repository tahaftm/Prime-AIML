```python
import requests
```

```python
res = requests.get("https://books.toscrape.com/")
```

```python
res.status_code
```

```python
if res.status_code == 200:
    with open("scraped_data/bookstore.html", "w") as f:
        f.write(res.text)
        print("Done writing the file")
```

### Headers in Web scrapping

Headers are used to depict that the request is comming from a real user not a python script

```python
headers = {
    "User-Agent" : "Mozilla/5.0"
}
```

```python
response = requests.get("https://books.toscrape.com/",headers=headers)
```
