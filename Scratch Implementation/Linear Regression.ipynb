```python
import pandas as pd
import numpy as np
```

```python
class LinearRegression():
    def __init__(self,alpha,n_iter):
        self.weight = None
        self.bias = None
        self.lr = alpha
        self.n_iter = n_iter

    def fit(self,X,y):
        m,n = X.shape

        ## Random initializing the thetas
        self.bias = 0
        self.weight = np.zeros(n)
        
        for i in range(1,self.n_iter):
            ## Developing the equation
            y_pred = self.bias + np.dot(X,self.weight)

            ## Derivate cost function(MSE) w.r.t bias
            db = (1/m)*(np.sum(y_pred - y))
            ## Derivate cost function(MSE) w.r.t Coefficients
            dw = (1/m)*(np.dot(X.T,y_pred - y))

            ## Convergence Theorem
            self.bias=self.bias - self.lr*db
            self.weight=self.weight- self.lr*dw

    def predict(self,X):
        y_pred = self.bias + np.dot(X,self.weight)
        return y_pred
```

```python
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])
```

```python
model = LinearRegression(alpha=0.1,n_iter=1000)
model.fit(X,y)
```

```python
model.predict([[1],[2],[3],[4],[5]])
```

### Ordinal Least Squares

- For small datasets

Technique: Normal Equation

```python
class LinearRegressionOLS():
    def __init__(self):
        self.bias = None
        self.weights = None
    def fit(self,X,y):
        m,n = X.shape
        X_b= np.c_[np.ones((m,1)),X]
        theta = np.dot(np.linalg.inv(np.dot(X_b.T,X_b)),np.dot(X_b.T,y))

        self.bias = theta[0]
        self.weights = theta[1:]
    def predict(self,X):
        y_pred = self.bias + np.dot(X,self.weights)
        return y_pred
```

```python
ols = LinearRegressionOLS()
ols.fit(X,y)
```

```python
ols.predict(X)
```
