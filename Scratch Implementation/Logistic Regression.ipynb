```python
import pandas as pd
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self,alpha= 0.1,n_iter = 1000):
        self.bias = None
        self.weights = None
        self.lr= alpha
        self.n_iter = n_iter

    def _sigmoid(self,pred):
        return (1/(1+np.exp(-pred)))
    
    def fit(self,X,y):
        m,n = X.shape
        self.bias = 0
        self.bias = 0
        self.weights = np.zeros(n)
        for i in range(1,self.n_iter):
            z = self.bias + np.dot(X,self.weights)

            y_pred = self._sigmoid(z)

            db = (1/m)*(y_pred - y)
            dw = (1/m)*(np.dot(X.T,y_pred - y))

            self.bias=self.bias - self.lr*db
            self.weights=self.weights- self.lr*dw

            self.bias=self.bias - self.lr*db
            self.weights=self.weights- self.lr*dw

    def predict_proba(self,X):
        z = self.bias + np.dot(X,self.weights)
        y_pred = self._sigmoid(z)
        return y_pred

    def predict(self,X):
        probablities= self.predict_proba(X)
        y_pred = probablities >= 0.5
        return y_pred.astype(int)


```

```python
X = np.array([[1,2],[2,3],[3,4],[4,5]])
y = np.array([0,0,1,1])
```

```python
model = LogisticRegression()
model.fit(X,y)
```

```python
model.predict(X)
```
